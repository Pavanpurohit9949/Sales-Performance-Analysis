```python
# Adding new columns
```

```python
df['profit_margin']=df['profit']/df['sales']
```

```python
df['shipping_days']=(df['ship_date'] - df['order_date']).dt.days
```

```python
# time features
```

```python
df['order_year'] = df['order_date'].dt.year
df['order_month']=df['order_date'].dt.month
df['order_year_month']=df['order_date'].dt.to_period('M').astype(str)
```

```python
# loss making orders
```

```python
df['is_loss']=df['profit']<0
```

```python
df[['sales', 'profit', 'discount', 'quantity']].describe()

# Flag extreme outliers worth knowing about before they skew a Power BI chart
df[df['profit'] < df['profit'].quantile(0.01)][['order_id', 'product_name', 'sales', 'profit', 'discount']]
```

```python
high_discount = df[df['discount'] >= 0.4]
low_discount = df[df['discount'] < 0.4]

print("Orders with discount >= 40%:")
print("  Count:", len(high_discount))
print("  Avg profit margin:", (high_discount['profit'].sum() / high_discount['sales'].sum() * 100).round(2), "%")

print("Orders with discount < 40%:")
print("  Count:", len(low_discount))
print("  Avg profit margin:", (low_discount['profit'].sum() / low_discount['sales'].sum() * 100).round(2), "%")
```

```python
import sqlalchemy
from sqlalchemy import create_engine

# Update these for your setup
server = 'localhost'          # or 'localhost\SQLEXPRESS' if using Express
database = 'SuperstoreDB'
driver = 'ODBC Driver 17 for SQL Server'

# Windows Authentication (most common for local dev)
conn_str = f'mssql+pyodbc://@{server}/{database}?driver={driver.replace(" ", "+")}&trusted_connection=yes'

engine = create_engine(conn_str)

df.to_sql('orders', con=engine, if_exists='replace', index=False)
```
